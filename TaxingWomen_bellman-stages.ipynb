# TaxingWomen Bellman Stages

## 1. The Paper and Its Model

### Paper analyzed; One-sentence summary

Guner, N., Kaygusuz, R., and Ventura, G. (2012), *Taxing women: A macroeconomic analysis*, *Journal of Monetary Economics* 59(2), 111-128.

The paper studies how taxes shape labor supply in a life-cycle model with singles and couples, with married female participation playing a central role because it affects both current household resources and future human capital.

### Agent's objective

Single males and single females maximize expected discounted lifetime utility; married households maximize discounted lifetime utility. The period utility functions are

For a single female, period utility depends on consumption, labor supply, and the extra time cost from young children when $k^y=1$.

$$
U_f^S(c,l,k^y)=\log(c)-\phi(l+k^y\kappa)^{1+1/\gamma},
$$

For a single male, period utility depends on consumption and labor supply only.

$$
U_m^S(c,l)=\log(c)-\phi l^{1+1/\gamma},
$$

For a married household, the wife's period utility includes her labor-disutility term and half of the fixed utility cost of joint work when she participates.

$$
U_f^M(c,l_f,q,k^y)=\log(c)-\phi(l_f+k^y\kappa)^{1+1/\gamma}-\frac{1}{2}\chi(l_f)q,
$$

The husband's period utility includes his own labor-disutility term, and the same fixed joint-work cost is split equally across the spouses.

$$
U_m^M(c,l_m,l_f,q)=\log(c)-\phi l_m^{1+1/\gamma}-\frac{1}{2}\chi(l_f)q.
$$

For married households, current utility is the sum of the two spouse-specific utilities.

$$
U_f^M(c,l_f,q,k^y)+U_m^M(c,l_m,l_f,q).
$$

### State variables at the start of a period

For single males, the state is $(a,z,j)$. For single females, the state is $(a,h,x,b,j)$. For married households, the state is $(a,h,x,z,q,b,j).$

Here:

- $a$ is assets.
- $h$ is current female human capital.
- $x$ is the female intrinsic type.
- $z$ is the male intrinsic type.
- $q$ is the utility cost of joint work.
- $b$ is childbearing status.
- $j$ is age.


### Control/choice variables within a period

- Single males choose labor supply $l$ and next period assets $a'$.
- Single females choose labor supply $l$ and next period assets $a'$.
- Married households choose female labor supply $l_f$, male labor supply $l_m$, and next period assets $a'$.
- In the recursive formulation, consumption $c$ is determined by the budget constraint once these choices are made.

### Shocks realized, and when

- Male type $z$.
- Female type $x$.
- Childbearing type $b$.
- For married households, the joint-work cost $q$.

These are realized at the start of life and then carried in the state. So in the paper's recursive problems, there is no new within-period shock between the choices.

### Bellman equation as the paper presents it

The paper's recursive problems are written as follows.

The single-male problem is

$$
V_m^S(a,z,j)=\max_{a',l}\left\{U_m^S(c,l)+\beta V_m^S(a',z,j+1)\right\},
$$

subject to

$$
c+a'=
\begin{cases}
a(1+r(1-\tau_k))+w\bar{\omega}_m(z,j)l(1-\tau_p)-T^S(w\bar{\omega}_m(z,j)l+ra,0), & j<J_R,\\
a(1+r(1-\tau_k))+p_m^S(z)-T^S(ra), & j\ge J_R,
\end{cases}
$$

and $l\ge 0$, $a'\ge 0$ (with strict equality if $j=J$).

The single-female problem is

$$
V_f^S(a,h,x,b,j)=\max_{a',l}\left\{U_f^S(c,l,k^y)+\beta V_f^S(a',h',x,b,j+1)\right\},
$$

subject to

$$
c+a'=
\begin{cases}
a(1+r(1-\tau_k))+whl(1-\tau_p)-T^S(whl+ra,1)-wd(j+1-b)\chi(l), & \text{with kids},\\
a(1+r(1-\tau_k))+whl(1-\tau_p)-T^S(whl+ra,0), & \text{without kids but not retired},\\
a(1+r(1-\tau_k))+p_f^S(x)-T^S(ra,0), & \text{retired},
\end{cases}
$$

with

$$
h' = G(x,h,l,j),
$$

and $l\ge 0$, $a'\ge 0$ (with strict equality if $j=J$).

The married-household problem is

$$
V^M(a,h,x,z,q,b,j)=\max_{a',l_f,l_m}\left\{\left[U_f^M(c,l_f,q,k^y)+U_m^M(c,l_m,l_f,q)\right]+\beta V^M(a',h',x,z,q,b,j+1)\right\}.
$$

subject to

$$
c+a'=
\begin{cases}
a(1+r(1-\tau_k))+w(\bar{\omega}_m(z,j)l_m+hl_f)(1-\tau_p)-T^M(w\bar{\omega}_m(z,j)l_m+whl_f+ra,1)-wd(j+1-b)\chi(l_f), & \text{with kids},\\
a(1+r(1-\tau_k))+w(\bar{\omega}_m(z,j)l_m+hl_f)(1-\tau_p)-T^M(w\bar{\omega}_m(z,j)l_m+whl_f+ra,0), & \text{without kids but not retired},\\
a(1+r(1-\tau_k))+p^M(x,z)-T^M(ra,0), & \text{retired},
\end{cases}
$$

with

$$
h' = G(x,h,l_f,j),
$$

and $l_m\ge 0$, $l_f\ge 0$, $a'\ge 0$ (with strict equality if $j=J$).

The decomposition below focuses on the married-household problem.

## 2. Why Multiple Stages?

The paper's married-household Bellman equation puts three controls inside one maximization: female labor, male labor, and savings. In a modular stage architecture, the question is not merely whether this problem can be split, but how to split it in a way that respects the economic timing.

For this model, the natural timing is straightforward. At the beginning of a working-age period the household knows its inherited state $(a,h,x,z,q,b,j)$. There is no new within-period uncertainty. The household then chooses female labor, chooses male labor, chooses savings (or equivalently consumption), and only after those choices does the model apply discounting to the next-period value. The next-period state is then carried through a connector into age $j+1$.

That logic suggests four within-period elements:

- a **female-labor stage**, because $l_f$ affects both current utility/resources and next period human capital;
- a **male-labor stage**, because $l_m$ affects current utility/resources but not the endogenous transition for human capital;
- a **consumption-savings stage**, because $a'$ is chosen only after within-period resources are pinned down;
- a **disc stage**, following `SolvingMicroDSOPs`, because discounting is conceptually separate from the economic choices.

### Timing used here

1. Start the period with inherited state $(a,h,x,z,q,b,j)$.
2. Choose female labor $l_f$.
3. Choose male labor $l_m$.
4. Choose next-period assets $a'$ and therefore consumption $c$.
5. Apply the discount factor to the next-period married value.
6. Use a connector to carry the post-decision state into the next age.

There is no new within-period shock between any of these steps.

### Why this helps

With that timing, each stage has one economically interpretable job:

- the **female-labor stage** handles the extensive/intensive female labor decision and the fixed utility cost of joint work;
- the **male-labor stage** adds the husband's labor decision without altering the next-period endogenous state;
- the **consumption-savings stage** converts current net resources into the post-decision state $(a',h')$;
- the **disc stage** houses the factor $\beta$ instead of burying it inside another stage's continuation object;
- the **connector** is a pure carry-forward/rename step between adjacent periods.

This decomposition is not the only possible one, but it is the cleanest modular representation of the paper's Bellman problem because it separates genuine economic choices from bookkeeping objects such as discounting and inter-period renaming.

## 3. Stage Decomposition

For the working-age married problem, I use the modular period

$$
\text{female-labor} \;\rightarrow\; \text{male-labor} \;\rightarrow\; \text{consumption-savings} \;\rightarrow\; \text{disc} \;\rightarrow\; [\text{connector}].
$$

The restriction to working age matters: this decomposition is for $j<J_R$. During working years the inherited state is $(a,h,x,z,q,b,j)$. The objects $x$, $z$, $q$, and $b$ are already known when the period begins, and there is no new uncertainty realized between the household's within-period choices.

Following `SolvingMicroDSOPs`, each stage is analyzed from three perches:

1. Arrival perch, indexed by $\prec$.
2. Decision perch, indexed by $\sim$.
3. Continuation perch, indexed by $\succ$.

I use $V_j^{M,\prec}$ for the period-level married value at age $j$, and I use $W$ for within-period stage values. Thus the first stage begins from

$$
V_j^{M,\prec}(a,h,x,z,q,b)=W_j^{f,\prec}(a,h,x,z,q,b).
$$

For compactness, once age $j$ is fixed I suppress $j$ from the stage-state arguments and let the period subscript carry the time index. This matches the spirit of the notation in `SolvingMicroDSOPs`, where the period label identifies the date and the stage state contains only the objects that are live inside the stage. Age still matters through age-dependent primitives such as $k^y(b,j)$, $\bar{\omega}_m(z,j)$, the child-care cost schedule $d(j+1-b)$, and the human-capital transition $G(x,h,l_f,j)$.

Because there are no new shocks between adjacent perches in this married working-age problem, the arrival-perch expectation operators are identities: for the choice stages, Arrival and Decision coincide. The only genuinely new operation after the three economic choices is discounting, which is why the notebook now treats discounting as its own stub stage rather than hiding $\beta$ inside the continuation value of the consumption stage.

### Stage 1: female-labor

**Name of the stage:** `female-labor`

| Perch | Indicator | State | value functions | Explanation |
| --- | --- | --- | --- | --- |
| Arrival | $\prec$ | $(a,h,x,z,q,b)$ | $W_j^{f,\prec}=W_j^{f,\sim}$ | At age $j$ the household arrives with the inherited married state. No new within-period shock arrives, so the arrival perch is already conditioned on $x,z,q,b$ and on the age-dependent child indicator $k^y(b,j)$. |
| Decision | $\sim$ | $(a,h,x,z,q,b)$ | $W_j^{f,\sim} = \max_{l_f\ge 0}\left\{-\phi(l_f+k^y\kappa)^{1+1/\gamma}-q\chi(l_f)+W_j^{f,\succ}\right\}$ | Choose female labor $l_f$. This is the economically distinctive stage because $l_f$ affects contemporaneous utility, participation, child-care use, and the next-period human-capital state. |
| Continuation | $\succ$ | $(a,h,x,z,q,b,l_f)$ | $W_j^{f,\succ}(a,h,x,z,q,b,l_f)=W_j^{m,\prec}(a,h,x,z,q,b,l_f)$ | The state passed to the next stage is the inherited state augmented by the chosen female labor supply. |

**Mathematical description:**

$$
W_j^{f,\sim}(a,h,x,z,q,b)
=
\max_{l_f\ge 0}
\left\{
-\phi(l_f+k^y\kappa)^{1+1/\gamma}
-q\chi(l_f)
+W_j^{f,\succ}(a,h,x,z,q,b,l_f)
\right\}.
$$

Since there is no uncertainty between the continuation perch of stage 1 and the arrival perch of stage 2,

$$
W_j^{f,\succ}(a,h,x,z,q,b,l_f)
=
W_j^{m,\prec}(a,h,x,z,q,b,l_f).
$$

This stage collects exactly the part of current utility that depends on $l_f$: the married female labor-disutility term and the fixed utility cost of joint work. It does **not** yet evaluate consumption utility or taxes; those belong downstream once the full within-period labor choice is known.

The human-capital transition induced by $l_f$ is

$$
h' = G(x,h,l_f,j)=\exp\!\big(\ln h + a_j^x\chi(l_f) - d(1-\chi(l_f))\big).
$$

### Stage 2: male-labor

**Name of the stage:** `male-labor`

| Perch | Indicator | State | value functions | Explanation |
| --- | --- | --- | --- | --- |
| Arrival | $\prec$ | $(a,h,x,z,q,b,l_f)$ | $W_j^{m,\prec}=W_j^{m,\sim}$ | The household arrives after the female labor choice has been fixed. Again, there is no new uncertainty between Arrival and Decision. |
| Decision | $\sim$ | $(a,h,x,z,q,b,l_f)$ | $W_j^{m,\sim} = \max_{l_m\ge 0}\left\{-\phi l_m^{1+1/\gamma}+W_j^{m,\succ}\right\}$ | Choose male labor $l_m$. This stage completes the within-period labor block but does not alter the endogenous human-capital transition. |
| Continuation | $\succ$ | $(a,h,x,z,q,b,l_f,l_m)$ | $W_j^{m,\succ}(a,h,x,z,q,b,l_f,l_m)=W_j^{c,\prec}(a,h,x,z,q,b,l_f,l_m)$ | The state passed forward records both labor choices, which is enough to evaluate current net resources in the savings stage. |

**Mathematical description:**

$$
W_j^{m,\sim}(a,h,x,z,q,b,l_f)
=
\max_{l_m\ge 0}
\left\{
-\phi l_m^{1+1/\gamma}
+W_j^{m,\succ}(a,h,x,z,q,b,l_f,l_m)
\right\}.
$$

Since there is no uncertainty between the continuation perch of stage 2 and the arrival perch of stage 3,

$$
W_j^{m,\succ}(a,h,x,z,q,b,l_f,l_m)
=
W_j^{c,\prec}(a,h,x,z,q,b,l_f,l_m).
$$

This stage collects exactly the part of current utility that depends only on $l_m$. Once $(l_f,l_m)$ are fixed, current labor income, taxes, and child-care expenditures can be evaluated in the next stage from the paper's budget constraint.

### Stage 3: consumption-savings

**Name of the stage:** `consumption-savings`

| Perch | Indicator | State | value functions | Explanation |
| --- | --- | --- | --- | --- |
| Arrival | $\prec$ | $(a,h,x,z,q,b,l_f,l_m)$ | $W_j^{c,\prec}=W_j^{c,\sim}$ | The household arrives after both labor choices have been fixed. No new shock occurs before the savings decision, so Arrival and Decision coincide. |
| Decision | $\sim$ | $(a,h,x,z,q,b,l_f,l_m)$ | $W_j^{c,\sim}=\max_{a'\ge 0}\left\{2\log c+W_j^{c,\succ}\right\}$ | Choose next-period assets $a'$, or equivalently choose consumption using the current budget constraint. |
| Continuation | $\succ$ | $(a',h',x,z,q,b)$ | $W_j^{c,\succ}(a',h',x,z,q,b)=W_j^{\beta,\prec}(a',h',x,z,q,b)$ | The post-decision state is the state that survives into the next period before discounting and before inter-period renaming. |

**Mathematical description:**

$$
W_j^{c,\sim}(a,h,x,z,q,b,l_f,l_m)
=
\max_{a'\ge 0}
\left\{
2\log c + W_j^{c,\succ}(a',h',x,z,q,b)
\right\},
$$

subject to

$$
c+a'=
\begin{cases}
a(1+r(1-\tau_k))+w(\bar{\omega}_m(z,j)l_m+hl_f)(1-\tau_p)-T^M(w\bar{\omega}_m(z,j)l_m+whl_f+ra,1)-wd(j+1-b)\chi(l_f), & \text{with kids},\\
a(1+r(1-\tau_k))+w(\bar{\omega}_m(z,j)l_m+hl_f)(1-\tau_p)-T^M(w\bar{\omega}_m(z,j)l_m+whl_f+ra,0), & \text{without kids but not retired}.
\end{cases}
$$

with

$$
h' = G(x,h,l_f,j).
$$

This is the stage at which the paper's resource constraint is evaluated. Once both labor choices are known, current net resources are pinned down and the only remaining economic choice is how much to carry into the next period through $a'$.

### Stage 4: disc

**Name of the stage:** `disc`

Following the modular architecture in `SolvingMicroDSOPs`, discounting is best treated as a stub stage rather than embedded inside the continuation value of the savings stage.

| Perch | Indicator | State | value functions | Explanation |
| --- | --- | --- | --- | --- |
| Arrival | $\prec$ | $(a',h',x,z,q,b)$ | $W_j^{\beta,\prec}=W_j^{\beta,\sim}$ | The disc stage inherits the post-decision state from the savings stage. |
| Decision | $\sim$ | $(a',h',x,z,q,b)$ | $W_j^{\beta,\sim}=\beta W_j^{\beta,\succ}$ | No economic choice occurs here; this stage simply applies discounting. |
| Continuation | $\succ$ | $(a',h',x,z,q,b)$ | $W_j^{\beta,\succ}(a',h',x,z,q,b)=V_{j+1}^{M,\prec}(C_j^M(a',h',x,z,q,b))$ | After discounting, the period hands off to the next period's arrival value through a connector. |

### Connector between periods

The connector is not a stage. It ties the continuation-perch state at the end of period $j$ to the arrival-perch state at the beginning of period $j+1$.

In the paper, the post-decision state at the end of the working-age period is $(a',h',x,z,q,b)$. At the beginning of the next period these same objects reappear as the inherited state $(a,h,x,z,q,b)$ one age later. The natural connector is therefore the pure rename

$$
C_j^M:(a',h',x,z,q,b) \leftrightarrow (a,h,x,z,q,b).
$$

Equivalently, one may read the connector as the assignments $a_{j+1}=a'_j$ and $h_{j+1}=h'_j$, with $x,z,q,b$ carried forward unchanged.

Using this connector, the end of the disc stage is

$$
W_j^{\beta,\succ}(a',h',x,z,q,b)
=
V_{j+1}^{M,\prec}(a',h',x,z,q,b),
$$

so that the disc stage itself is

$$
W_j^{\beta,\sim}(a',h',x,z,q,b)
=
\beta V_{j+1}^{M,\prec}(a',h',x,z,q,b).
$$

### Composition of stages

The full modular working-age period is therefore

$$
[\text{female-labor},\; \text{male-labor},\; \text{consumption-savings},\; \text{disc}].
$$

In forward timing, the flow is

$$
\text{female-labor} \;\rightarrow\; \text{male-labor} \;\rightarrow\; \text{consumption-savings} \;\rightarrow\; \text{disc} \;\rightarrow\; [\text{connector}].
$$

In backward solution order, the flow is

$$
[\text{connector}] \;\leftarrow\; \text{disc} \;\leftarrow\; \text{consumption-savings} \;\leftarrow\; \text{male-labor} \;\leftarrow\; \text{female-labor}.
$$

The modular Bellman representation is

$$
V_j^{M,\prec}(a,h,x,z,q,b)
=
W_j^{f,\prec}(a,h,x,z,q,b)
=
W_j^{f,\sim}(a,h,x,z,q,b),
$$

with

$$
W_j^{f,\sim}(a,h,x,z,q,b)
=
\max_{l_f\ge 0}
\left\{
-\phi(l_f+k^y\kappa)^{1+1/\gamma}-q\chi(l_f)
+W_j^{m,\prec}(a,h,x,z,q,b,l_f)
\right\},
$$

$$
W_j^{m,\prec}(a,h,x,z,q,b,l_f)
=
W_j^{m,\sim}(a,h,x,z,q,b,l_f)
=
\max_{l_m\ge 0}
\left\{
-\phi l_m^{1+1/\gamma}
+W_j^{c,\prec}(a,h,x,z,q,b,l_f,l_m)
\right\},
$$

$$
W_j^{c,\prec}(a,h,x,z,q,b,l_f,l_m)
=
W_j^{c,\sim}(a,h,x,z,q,b,l_f,l_m)
=
\max_{a'\ge 0}
\left\{
2\log c + W_j^{\beta,\prec}(a',G(x,h,l_f,j),x,z,q,b)
\right\},
$$

and

$$
W_j^{\beta,\prec}(a',h',x,z,q,b)
=
W_j^{\beta,\sim}(a',h',x,z,q,b)
=
\beta V_{j+1}^{M,\prec}(a',h',x,z,q,b).
$$

Stacking these four elements reproduces the paper's original Bellman problem exactly; the decomposition only makes the within-period logic explicit.

### Retirement

For $j\ge J_R$ the labor choices disappear from the paper's problem. In modular terms, the period collapses from

$$
[\text{female-labor},\; \text{male-labor},\; \text{consumption-savings},\; \text{disc}]
$$

to a simpler retired period with only savings and discounting, followed by the same inter-period connector.

## 4. Why This Decomposition Is Better

The original notebook had the right broad intuition but it mixed together two different goals: faithfulness to the paper's Bellman problem and faithfulness to the modular architecture of `SolvingMicroDSOPs`. In particular, it embedded discounting inside the continuation value of the savings stage, used perch notation even when the economic role of each perch was not yet clear, and blurred the distinction between true stage objects and the connector that links adjacent periods.

The revised decomposition is better for three reasons.

First, it is more faithful to the economics of the paper. The married household really solves one working-age problem with three controls $(l_f,l_m,a')$, no new within-period shocks, and a transition in which only female labor changes next period human capital. The stage order now mirrors exactly that logic.

Second, it is more faithful to the modular architecture in `SolvingMicroDSOPs`. The notebook now treats discounting as its own stub stage, and it treats the connector as a pure inter-period rename/carry-forward object rather than as an informal afterthought inside the savings stage.

Third, it is more useful for extension. If one later wanted to add a participation stage, a tax-distortion stage, a shock-realization stage, or alternative timing assumptions, the present representation makes clear where those modifications would enter and which existing stages would remain untouched.

There are still alternative decompositions one could defend. For example, one could combine the two labor choices into one joint labor stage. That would also be faithful to the paper. I do not adopt that representation here because it hides the asymmetry that matters most in this model: female labor changes the future endogenous state through $G(x,h,l_f,j)$, while male labor does not. Splitting the labor block into female and male stages therefore gives the cleanest modular interpretation of the paper's economics.
